```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer
x=[[1,2,np.nan],[3,np.nan,3],[np.nan,60,5],[8,8,7]] 
impute = KNNImputer(n_neighbors=2,
                   weights='uniform')   # all obesrvation is equle 
impute.fit_transform(x)
```

```python
 import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer
x=[[1,2,np.nan],[3,np.nan,3],[np.nan,60,5],[8,8,7]] 
impute = KNNImputer(n_neighbors=2,
                   weights='distance')
impute.fit_transform(x)
```
