## fill the outliers

- drop the outlier

- fill with median

- winsonization fill LB and UB

```python
con1 = wage_data > ub
con2 = wage_data < lb
con = con1 | con2 
# We will the iterate the wach value frome wagw data
# if the condition matches 
```

```python
outlier_df = visa_df[con]

outlier = outlier_df['prevailing_wage'].values
wage_data = visa_df['prevailing_wage']
wage_median = wage_data.median()
l = []

for value in wage_data:
    if value in outlier:
        l.append(wage_median)
    else:
        l.append(value)

visa_df['prevailing_wage1']=l
visa_df
```
